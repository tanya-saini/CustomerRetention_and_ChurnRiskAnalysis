# Retail Sales & Customer Analytics

## Project Objective

Analyze online retail transaction data to uncover sales, product, geographic, and customer behavior patterns; perform RFM-based customer segmentation to identify valuable and at-risk customer groups; and translate analytical findings into actionable recommendations for customer retention, revenue growth, and business performance.

RFM analysis will be used to identify customer segments based on:

- Recency
- Frequency
- Monetary value

---

## Key Business Questions

### 1. How is the business performing?

- What is total revenue?
- How many orders and units are being sold?
- How does revenue change over time?
- What is the average order value?

### 2. What products and markets drive revenue?

- Which products generate the most revenue?
- Which products sell the most units?
- How concentrated is revenue among products?
- Which countries are the most important markets?

### 3. Who are the most valuable customers?

- Who purchases most frequently?
- Who generates the most revenue?
- Who are the recent customers?
- Which customers are loyal?
- Which customers appear to be at risk?

### 4. How can the business act on these findings?

- Which customers should receive retention offers?
- Which customer segments deserve targeted marketing?
- Which products should receive more attention?
- Which markets are most important?
- Where are opportunities for growth?
